```python
import numpy as np

#Simulating crop yield data for 6 states
states = ["Punjab", "Maharashtra", "UP", "MP", "Rajasthan", "Karnataka"]
wheat_yield = np.array([4800, 1200, 3200, 2800, 900, 800])
rice_yield = np.array([2100, 3800, 4200, 1900, 400, 3100])

total_yield = wheat_yield + rice_yield

print("--- Crop Yield Analysis --- \n")

print(f"Average wheat yield: {np.mean(wheat_yield):.2f} kg")
print(f"Average rice yield: {np.mean(rice_yield):.2f} kg")
print(f"Highest wheat yield: {np.max(wheat_yield)} kg")
print(f"Lowest rice yield: {np.min(rice_yield)} kg")

index = np.argmax(total_yield) #Gives index of largest value

print(f"\nBest performing state: {states[index]}")
print(f"Total yield: {total_yield[index]} kg")

#States above average total yield

avg = np.mean(total_yield)
print(f"\nAverage total yield: {avg:.2f} kg \n") #IMP FORMATTING

print("States above average:")
for i, (states, total) in enumerate(zip(states, total_yield)):
    if total >= avg:
        print(f"{states}: {total} kg ")


# Yield in tons
print(f"\nWheat yield in tons: {wheat_yield / 1000}")
print(f"Rice yield in tons: {rice_yield / 1000}")
```
