```python
import pandas as pd
import numpy as np

# --- Messy data (like real government data) ---
data = {
    "state": ["Punjab", "Maharashtra", None, "MP", "Rajasthan", "Karnataka", "Bihar", None],
    "crop": ["Wheat", "sugarcane", "Rice", "SOYBEAN", "Bajra", "ragi", None, "Cotton"],
    "yield_kg": [4800, 8200, None, 1200, -999, 800, 2100, 3800],
    "year": [2023, 2023, 2022, 2023, 2022, 2023, None, 2023],
    "area_hectares": [3200, 1800, 4100, 2900, 1500, 600, 2200, 3100]
}

df = pd.DataFrame(data)
print("Original messy data:")
print(df)
print("\nNull values per column:")
print(df.isnull().sum())

# --- Fix 1: Drop rows where state is null ---
df = df.dropna(subset=["state"])
print("\nAfter dropping null states:")
print(df)

# --- Fix 2: Fill null yield with mean ---
mean_yield = df["yield_kg"].mean()
df["yield_kg"] = df["yield_kg"].fillna(mean_yield)
print(f"\nFilled null yield with mean: {mean_yield:.2f}")

# --- Fix 3: Fill null year with most common year ---
df["year"] = df["year"].fillna(df["year"].mode()[0])

# --- Fix 4: Fix negative yield values ---
df["yield_kg"] = df["yield_kg"].apply(lambda x: mean_yield if x < 0 else x)
print("\nAfter fixing negative values:")
print(df["yield_kg"])

# --- Fix 5: Standardize text (uppercase, strip spaces) ---
df["crop"] = df["crop"].str.strip().str.title()
df["state"] = df["state"].str.strip().str.title()
print("\nAfter standardizing text:")
print(df[["state", "crop"]])

# --- Fix 6: Drop null crops ---
df = df.dropna(subset=["crop"])

# --- Final clean data ---
print("\n✅ Final cleaned data:")
print(df)
print(f"\nShape: {df.shape}")
print(f"Remaining nulls: {df.isnull().sum().sum()}")

# Save clean data
df.to_csv("cleaned_crop_data.csv", index=False)
print("\n✅ Saved to cleaned_crop_data.csv")
```
