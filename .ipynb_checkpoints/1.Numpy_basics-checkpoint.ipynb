```python
# Import and array creation
import numpy as np


arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10, 20, 30, 40, 50])

print("Array 1:", arr1)
print("Array 2:", arr2)
print("Type:", type(arr1))
print("Shape:", arr1.shape)
print("Dtype:", arr1.dtype)
```

```python
# Array Mathematics
print(f"Add 10 to every element: {arr1 + 10}")
print(f"Multiply by 2: {arr1 * 2}")
print(f"arr1 + arr2: {arr1 + arr2}")
print(f"arr1 * arr2: {arr1 * arr2}")
```

```python
# Statistics
yields = np.array([4800, 8200, 3100, 1200, 900, 3900, 5200, 2100])

print(f"Mean: {np.mean(yields)}")
print(f"Max: {np.max(yields)}")
print(f"Min: {np.min(yields)}")
print(f"Sum: {np.sum(yields)}")
print(f"Std deviation: {np.std(yields)}")
print(f"Median: {np.median(yields)}")
```

```python
# Filtering

avg = np.mean(yields)
print(f"Average yield: {avg}")

high_yield = yields[yields >= avg]
print(f"Above average yields: {high_yield}")

low_yield = yields[yields < avg]
print(f"Below average yields: {low_yield}")
```

```python
# Reshaping

matrix = yields.reshape(2,4)
print(f"Reshaped to 2x4 matrix: \n{matrix}")
print(f"Shape: {matrix.shape}")
```

```python
#Useful array creations

zeros = np.zeros(5)
ones = np.ones(5)
range_arr = np.arange(0, 100, 10) # 0 to 100, step of 10
random_arr = np.random.randint(1000,9000,size = 10) # 10 random number

print("Zeros:", zeros)
print("Ones:", ones)
print("Range:", range_arr)
print("Random:", random_arr)
```
