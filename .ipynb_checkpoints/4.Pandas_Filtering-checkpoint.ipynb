```python
import pandas as pd

data = {
    "state": ["Punjab", "Maharashtra", "UP", "MP", "Rajasthan", "Karnataka", "Bihar", "Gujarat"],
    "crop": ["Wheat", "Sugarcane", "Rice", "Soybean", "Bajra", "Ragi", "Maize", "Cotton"],
    "yield_kg": [4800, 8200, 3100, 1200, 900, 800, 2100, 3800],
    "year": [2023, 2023, 2022, 2023, 2022, 2023, 2022, 2023],
    "area_hectares": [3200, 1800, 4100, 2900, 1500, 600, 2200, 3100]
}

df = pd.DataFrame(data)

# --- Selecting columns ---
print("Single column:")
print(df["state"])

print("\nMultiple columns:")
print(df[["state", "yield_kg"]])

# --- Filtering rows ---
print("\nYield above 3000:")
print(df[df["yield_kg"] >= 3000])

print("\nYear 2023 only:")
print(df[df["year"] == 2023])

print("\nYear 2023 AND yield above 2000:")
print(df[(df["year"] == 2023) & (df["yield_kg"] >= 2000)])

print("\nSpecific states:")
print(df[df["state"].isin(["Punjab", "Maharashtra", "Gujarat"])])

# --- Sorting ---
print("\nSorted by yield (highest first):")
print(df.sort_values("yield_kg", ascending=False))

# --- Adding new column ---
df["yield_tons"] = df["yield_kg"] / 1000
df["above_average"] = df["yield_kg"] > df["yield_kg"].mean()
print("\nWith new columns:")
print(df)
```
