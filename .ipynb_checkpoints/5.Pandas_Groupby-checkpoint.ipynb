```python
import pandas as pd

data = {
    "state": ["Punjab", "Punjab", "Maharashtra", "Maharashtra", "UP", "UP", "MP", "MP"],
    "crop": ["Wheat", "Rice", "Sugarcane", "Cotton", "Rice", "Wheat", "Soybean", "Wheat"],
    "yield_kg": [4800, 2100, 8200, 3800, 3100, 2900, 1200, 1800],
    "year": [2023, 2023, 2023, 2022, 2023, 2022, 2023, 2022],
    "area_hectares": [3200, 1800, 1800, 2200, 4100, 3000, 2900, 2100]
}

df = pd.DataFrame(data)

# --- Basic groupby ---
print("Average yield per state:")
print(df.groupby("state")["yield_kg"].mean())

print("\nTotal yield per state:")
print(df.groupby("state")["yield_kg"].sum())

print("\nMax yield per crop:")
print(df.groupby("crop")["yield_kg"].max())

# --- Multiple aggregations ---
print("\nMultiple stats per state:")
print(df.groupby("state").agg({
    "yield_kg": ["mean", "max", "min"],
    "area_hectares": "sum"
}))

# --- Groupby year ---
print("\nAverage yield per year:")
print(df.groupby("year")["yield_kg"].mean())

# --- Count per state ---
print("\nNumber of crops per state:")
print(df.groupby("state")["crop"].count())
```
