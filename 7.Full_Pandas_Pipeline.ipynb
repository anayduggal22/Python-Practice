```python
import pandas as pd
import numpy as np

# --- Step 1: Load data ---
# Simulate loading from a CSV (like a government dataset)
data = {
    "state": ["Punjab", "Maharashtra", "UP", "MP", "Rajasthan",
              "Karnataka", "Bihar", "Gujarat", "Punjab", "Maharashtra"],
    "crop": ["Wheat", "Sugarcane", "Rice", "Soybean", "Bajra",
             "Ragi", "Maize", "Cotton", "Rice", "Cotton"],
    "yield_kg": [4800, 8200, 3100, 1200, 900, 800, 2100, 3800, 2200, 4100],
    "year": [2023, 2023, 2022, 2023, 2022, 2023, 2022, 2023, 2023, 2022],
    "area_hectares": [3200, 1800, 4100, 2900, 1500, 600, 2200, 3100, 1900, 2800]
}
df = pd.DataFrame(data)
print(f"✅ Loaded {len(df)} records")

# --- Step 2: Explore ---
print(f"\nShape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"Nulls:\n{df.isnull().sum()}")

# --- Step 3: Clean ---
df["crop"] = df["crop"].str.strip().str.title()
df["state"] = df["state"].str.strip().str.title()
df = df.dropna()
df = df[df["yield_kg"] > 0]
print(f"\n✅ After cleaning: {len(df)} records")

# --- Step 4: Analyse ---
print("\n=== Analysis ===")
print(f"Average yield: {df['yield_kg'].mean():.2f} kg")
print(f"Best state: {df.groupby('state')['yield_kg'].mean().idxmax()}")
print(f"Best crop: {df.groupby('crop')['yield_kg'].mean().idxmax()}")

print("\nAverage yield per state:")
state_avg = df.groupby("state")["yield_kg"].mean().sort_values(ascending=False)
print(state_avg)

# --- Step 5: Filter ---
above_avg = df[df["yield_kg"] >= df["yield_kg"].mean()]
print(f"\nRecords above average yield: {len(above_avg)}")

# --- Step 6: Save ---
df.to_csv("final_pipeline_output.csv", index=False)
print("\n✅ Saved to final_pipeline_output.csv")
print("Pipeline complete!")
```
